```python
dataset_path = "dt_creado"
categorias = ["arboles", "casas", "puentes"]

imagenes_lista = []
tEsperado = (80, 80)

for categoria in categorias:
    carpeta_cat = os.path.join(dataset_path, categoria)
    
    if not os.path.exists(carpeta_cat):
        print(f"Advertencia: La carpeta {carpeta_cat} no existe.")
        continue
        
    print(f"Procesando categoría: {categoria}")
    
    for archivo in os.listdir(carpeta_cat):
        ruta_img = os.path.join(carpeta_cat, archivo)
        
        # Asegurarse de que sea un archivo
        if os.path.isfile(ruta_img):
            try:
                with Image.open(ruta_img) as img:
                    # Verificar dimensiones exactas de 80x80
                    if img.size != tEsperado:
                        raise ValueError(f"La imagen {archivo} tiene tamaño {img.size}, se esperaba {tEsperado}.")
                    
                    # Opcional: Asegurarse de que sea a color (RGB)
                    img = img.convert("RGB")
                    
                    # Convertir a arreglo numpy
                    img_array = np.array(img)
                    imagenes_lista.append(img_array)
            except Exception as e:
                print(f"No se pudo procesar {archivo}: {e}")

print(f"\n¡Verificación completada! Total de imágenes cargadas correctamente: {len(imagenes_lista)}")
```
