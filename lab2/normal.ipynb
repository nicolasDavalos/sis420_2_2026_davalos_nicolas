metodo de la ecuacion de la normal

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as pyplot
```

```python
data = pd.read_csv(
    'dataset_2/Loan (1).csv'
)

print("Número de ejemplos:", data.shape[0])
print("Número de columnas:", data.shape[1])

print(data.head())
```

```python
# Seleccionar únicamente las columnas numéricas
data_numeric = data.select_dtypes(include=np.number)

# X contiene las variables predictoras
X = data_numeric.drop(columns=['RiskScore'])

# y contiene la variable que queremos predecir
y = data_numeric['RiskScore']

# Convertir a matrices NumPy
X = X.to_numpy()
y = y.to_numpy()

# Número de ejemplos
m = y.size

print("Número de ejemplos:", m)
print("Número de variables:", X.shape[1])
print("Dimensiones de X:", X.shape)
print("Dimensiones de y:", y.shape)
```

```python
def featureNormalize(X):

    X_norm = X.copy()

    mu = np.mean(X, axis=0)
    sigma = np.std(X, axis=0)

    X_norm = (X - mu) / sigma

    return X_norm, mu, sigma
```

```python
X_norm, mu, sigma = featureNormalize(X)

print("Media:")
print(mu)

print("\nDesviación estándar:")
print(sigma)

print("\nDimensiones de X_norm:")
print(X_norm.shape)
```

```python
X = np.concatenate([np.ones((m, 1)), X_norm], axis=1)

print("Dimensiones de X:", X.shape)
print(X[:5])
```

```python
def normalEqn(X, y):

    theta = np.zeros(X.shape[1])

    theta = np.linalg.pinv(X.T @ X) @ X.T @ y

    return theta
```

```python
# Calcula los parámetros con la ecuación normal
theta_normal = normalEqn(X, y)

print("Theta calculado mediante la ecuación normal:")
print(theta_normal)
```

```python
def computeCostMulti(X, y, theta):

    m = y.shape[0]

    J = (1 / (2 * m)) * np.sum(
        np.square(np.dot(X, theta) - y)
    )

    return J
```

```python
J_normal = computeCostMulti(X, y, theta_normal)

print("Costo utilizando la ecuación normal:", J_normal)
```

```python
# Seleccionamos el primer ejemplo
X_ejemplo = X[10]

# Realizamos la predicción
prediccion = np.dot(X_ejemplo, theta_normal)

# Valor real
valor_real = y[10]

print("RiskScore real:", valor_real)
print("RiskScore predicho:", prediccion)
print("Error:", prediccion - valor_real)
```
